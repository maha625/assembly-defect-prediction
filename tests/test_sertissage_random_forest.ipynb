# Test d'une courbe de sertissage avec Random Forest
Ce notebook charge le modèle optimisé (`crimp_best_random_forest_model.pkl`), extrait les caractéristiques d'une nouvelle courbe brute et prédit si la pièce est conforme (**OK**) ou défectueuse (**NOK**).

## 1. Importation des bibliothèques

```python
import joblib
import numpy as np
import pandas as pd
```

## 2. Chargement du modèle Random Forest entraîné

```python
model_filename = "./../models/crimp_best_random_forest_model.pkl"
rf_model = joblib.load(model_filename)
print("Modèle Random Forest chargé avec succès.")
```

## 3. Définition ou chargement d'une nouvelle courbe brute

```python
# Exemple de courbe brute (remplacer par les valeurs d'une acquisition capteur réelle)
nouvelle_courbe_brute = [12.5, 14.2, 18.0, 25.6, 45.1, 78.3, 95.0, 85.2, 50.4, 30.1, 15.0]
print(f"Nombre de points dans la courbe : {len(nouvelle_courbe_brute)}")
```

## 4. Extraction des caractéristiques (Feature Engineering)

```python
arr = np.array(nouvelle_courbe_brute)
features = [
    np.max(arr),       # Force maximale atteinte
    np.min(arr),       # Force minimale
    np.mean(arr),      # Force moyenne
    np.std(arr),       # Écart-type (variation du signal)
    np.trapz(arr),     # Aire sous la courbe (énergie de sertissage)
    np.argmax(arr)     # Position temporelle du pic de force
]

# Formatage en tableau 2D pour le modèle
X_input = np.array([features])
print("Caractéristiques extraites :", X_input)
```

## 5. Prédiction et affichage des résultats

```python
prediction = rf_model.predict(X_input)[0]
probabilities = rf_model.predict_proba(X_input)[0]

print("--- Résultat du test (Random Forest) ---")
print(f"Statut prédit : {'OK (Conforme)' if prediction == 0 else 'NOK (Anomalie)'}")
```
